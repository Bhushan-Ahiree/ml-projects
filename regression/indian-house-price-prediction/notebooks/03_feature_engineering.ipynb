```python
import joblib
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.read_csv("../data/processed/House_Price_Data_cleaned.csv")

df.head()
```

```python
x = df.drop(columns=["totalprice"])

y = df["totalprice"]
```

```python
print(x.head())

print(y.head())
```

```python
categorical_features = [
    "propertytype",
    "location",
]

numerical_features = [
    "bhk",
    "sqft",
]
```

Why are we doing this?

Because different features need different preprocessing.

```python
categorical_transformer = OneHotEncoder(
    handle_unknown="ignore"
)
```

Why ignore?

Training data

Pune
Mumbai
Nashik

Prediction

Nagpur

Without ignore

↓

Error

With ignore

↓

Prediction still works.

Production systems almost always use this.

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            categorical_transformer,
            categorical_features,
        )
    ],
    remainder="passthrough",
)
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

Why 42?

Not magic.

Just reproducibility.

Every time you run

You get the same split.

Useful for debugging and experiments.

```python
X_train_processed = preprocessor.fit_transform(X_train)
```

Why fit?

Because the encoder must learn

Apartment

Villa

Row House

Only from training data.

Never from testing data.

Otherwise

↓

Data Leakage.

```python
X_test_processed = preprocessor.transform(X_test)
```

```python
print(X_train_processed.shape)

print(X_test_processed.shape)
```

```python
joblib.dump(
    preprocessor,
    "../models/preprocessors/preprocessor.pkl",
)
```
