# 🧹 Data Preprocessing

## Objective

Prepare the dataset for Machine Learning by:

- Removing duplicate records
- Separating features and target
- Handling categorical variables
- Splitting training and testing data

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.read_csv("../data/raw/House_Price_Data.csv")

df.head()
```

```python
df.shape
```

```python
df.duplicated().value_counts()
```

Check the dataset size before preprocessing.

```python
df = df.drop_duplicates()

df.shape
```

```python
df = df.drop(columns=["pricepersqft"])

df.columns
```

The `pricepersqft` feature is excluded because it is directly related to the target (`totalprice`) and may introduce data leakage.

```python
x = df.drop(columns=["totalprice"])

y = df["totalprice"]
```

```python
x.head()
```

```python
y.head()
```

```python
categorical_features = ["propertytype", "location"]

numerical_features = ["bhk", "sqft"]
```

Numerical and categorical features require different preprocessing techniques.

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features,
        )
    ],
    remainder="passthrough",
)
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)
print("Training Labels   :", y_train.shape)
print("Testing Labels    :", y_test.shape)
```

The dataset has been split into 80% training data and 20% testing data.

```python
X_train_processed = preprocessor.fit_transform(X_train)     # Apply Preprocessing

X_test_processed = preprocessor.transform(X_test)
```
