# 🧹 Data Cleaning

## Objective

Prepare a clean dataset before feature engineering.

Tasks:

- Remove duplicate records
- Inspect data types
- Detect outliers
- Remove unnecessary columns
- Save cleaned dataset

```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/House_Price_Data.csv")

df.head()
```

```python
print(df.shape)
```

```python
duplicates = df.duplicated().sum()

print("Duplicate Rows :", duplicates)
```

```python
df = df.drop_duplicates()

print(df.shape)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.nlargest(10, "sqft")
```

```python
df.nsmallest(10, "sqft")
```

```python
df = df.drop(columns=["pricepersqft"])

df.head()
```

## Why remove `pricepersqft`?

Since

Price ≈ Area × Price per sqft

using this feature may leak information about the target variable.

The first model will therefore exclude this feature.

```python
df.to_csv(
    "../data/processed/House_Price_Data_cleaned.csv",
    index=False,
)
```

```python
cleaned = pd.read_csv("../data/processed/House_Price_Data_cleaned.csv")

cleaned.head()
```

# Summary

Completed

- Removed duplicate records
